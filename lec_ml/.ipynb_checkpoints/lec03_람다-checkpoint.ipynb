# **컨프리헨션**

* for문 1~10까지 짝수만 출력

```python
for a in range (1, 11) :
    if ( a % 2 == 0) :
        print(a, end="  ")
```

* 컨프리헨션

```python
[ a for a in range (1, 11) if ( a % 2 == 0) ]
```

* for문 1~5까지 3보다 크면 값*2, 그렇지 않으면 값 그대로

```python
for a in range (1, 6) :
    if ( a > 3) :
        print(a*2, end="  ")
    else :
        print(a, end="  ")
```

* 컨프리헨션

```python
[ a*2 if ( a > 3) else a for a in range (1, 6) ]
```

# **람다(Lambda)**
* 익명 함수 (이름이 없는)
* 1회용

* 함수이름 : add
* 파라미터 : 숫자 2개
* 리턴 : 합한 결과

```python
def add(n1, n2) : 
    res = n1 + n2
    return res
```

```python
a = add(4, 5)
print(a)
```

```python
def ncheck(n) :
    print(n*2)

ncheck(6)
```

* <b> 람다 예제

```python
ncheck2 = lambda n : n*2
```

```python
ncheck2(6)
```

* <b> 람다 예제2

```python
def add(n1, n2) : 
    res = n1 + n2
    return res
    
a = add(4, 5)
print(a)
```

```python
add2 = lambda n1, n2 : n1 + n2
add2(4, 5)
```

* <b> 람다 예제3 + 컨프리헨션

```python
a = lambda n : n if n > 10 else 0
a (3)
```

# **DataFrame을 활용한 람다(Lambda)**

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "name": ["kim", "hong", "park"],
    "age": [20, 25, 30]
})
df
```

```python
df["age"] = df["age"] + 10
df
```

```python
df["age"] = df["age"].apply(lambda a : a+10)
df
```

```python
df["etc"] = df["age"].apply(lambda a : '성인' if a>=45 else '미성년')
df
```
