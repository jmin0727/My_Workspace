# **Lec05. [실습] 주소록**

## **실습**

```python
addr_list = []   #처음 주소록에 아무것도 없다

while (True) :   #일단 트루니까 반복 시작
    print("________________________")
    print("1:등록  2:수정   3:삭제   4:조회   Q:종료")
    menu = input("메뉴를 선택하세요")    #메뉴는 인풋으로 직접 입력 (아래 조건에 해당하면 시작)

    if (menu in ["Q", "q"]) : 
        print("종료")
        break
        
    elif (menu == "1" ): 
        name = input("이름을 입력하세요")
        tel = input("전화번호를 입력하세요")
        
        addr_list.append( [name, tel] )         #주소록에 추가
        print(f"{name}, {tel} 저장되었습니다")
    
    elif (menu == "2" ):
        print("수정")
        search_name = input("수정하려면 이름을 입력하세요")   #새 변수 = 사용자 입력
        update_tel = input("수정할 전화번호를 입력하세요")    #새 변수 = 사용자 입력
        for v in addr_list :                  #v는 항상 덩어리로 나옴
            if ( v[0] == search_name ) :      #인덱스 사용 신규값 할당 (덩어리 중 0번째로 서치)
                v[1] = update_tel
    
    elif (menu == "3" ):
        print("삭제")
        search_name = input("삭제할 사람의 이름을 입력하세요")
        for v in addr_list :
            if ( v[0] == search_name ) :     #v의 0번째가 사용자가 입력한 값과 같으면 (인덱스 사용)
                addr_list.remove(v)          #값 전체 삭제 
    
    elif (menu == "4" ):
        print(f"{len( addr_list )}건 조회")
        for v in addr_list :
            print(v[0], v[1])
    
    else :
        print(f"잘못된 메뉴 번호를 선택하셨습니다.")
```
