````python
for a in range(1, 6) :
    print(a, end=" ")
````

* 1부터 5까지 for문 사용해 총합 출력

````python
total = 0
for a in range(1, 6) :
   total = total + a
print(total)
````

# **제어문 : while**

* 1부터 5까지 while문 사용해 출력

````python
num = 1
while ( num <= 5) :
    print (num, end=" ")
    num = num + 1
````

````python
num = 1
while (True) :
    print(num, end=" ")
    
    if (num == 5):
        break
    num = num + 1
````

````python
num = 1
while (True) :
    if (num > 5) :
        break
    else :
        print(num, end=" ")
        num = num + 1
````

* 숫자를 입력받아 합의 값을 돌려주는 함수
  * 함수이름 : calc_point2
  * 파라미터 : num1, num2
  * 기능 : num1 2배값과 num2 10배값의 합
  * 리턴 : 계산값

````python
def calc_point2(num1, num2) :
    return num1 * 2 + num2 * 10
    
a = calc_point2(4, 5)
print(a)
````

* 출석부에 학생 이름을 입력하면 추가하는 함수
