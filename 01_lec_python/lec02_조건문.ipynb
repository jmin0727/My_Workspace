# **Lec02. 조건문**

> 조건 : 참(True), 거짓(False) = 불리언(bool) 타입만 가능

**if문**
```python
if (조건)
    코드1
    코드2
----------------
if (조건) :
    코드1
    코드2
else :
    코드3   #if조건에 맞지 않다면(조건 외 나머지라면 코드3)
----------------
if (조건) :  #조건1
    코드1
    코드2
elif (조건) :
    코드3
    코드4   #조건2
else :
    코드5   #나머지, else는 조건이 없다

**<u>중첩(멀티)</u> if문 : 디테일 Up**
``` python
if (조건) :
    코드1
    if (조건) :
        코드2
    elif :
        코드3
else :
    코드4

## **조건문**

````python
score = 90
print( score >= 95 )
````

````python
if (score >= 95) :
    print ("aaa")    #score가 95 이상이 아니기 때문에 아무것도 출력 안 됨
````

## **멀티 조건문**

* <font color=blue> and : 모든 조건 다 True </font>

````python
age = 25
gen = 'm'
age < 30 and gen == 'm'
````

````python
age = 60
gen = 'm'
age < 30 and gen == 'm'
````

````python
age = 25
gen = 'm'
height = 148
age < 30 and gen =='m' and height >=150
````

* <font color=blue> or : 조건들 중 하나라도 True

````python
age = 60
gen = 'm'
age < 30 or gen == 'm'
````

* <font color=blue> in : 리스트 안에 값이 있는가

````python
3 in [1, 2, 3]
````

````python
mylist = [1, 2, 3]
3 in mylist
````

* <font color=blue> not : 리스트 안에 값이 없는가

````python
mylist = [1, 2, 3]
3 not in mylist
````

* <font color=blue> == : 같다

````python
a = 10
b = 20
a == b
````

* <font color=blue> != : 다르다

````python
a = 10
b = 20
a != b
````

````python
a = 10
b = 20

if (a != b) :
    print("다르다")
else :
    print("같다")
````

## **if문법**

````python
score = 90
if (score >=90 ) :
    print("A")
````

````python
score = 85
if (score >=90 ) :
    print("A")
else :
    print("나머지")
````

````python
score = 85
if (score >=90 ) :
    print("A")
else :
    print("나머지")
````

````python
score = 95
if (score >=90 ) :
    print("A")
elif (score >=80 ) :
    print("B")
elif (score >=70) :
    print("C")
else :
    print("나머지")
````

````python
score = 85
if (score >=90 ) :
    print("A")
elif (score >=80 ) :
    print("B")
elif (score >=70) :
    print("C")
else :
    print("나머지")

print("---done---")   #C에도 부합하지만 B 우선조건이 맞기 때문에 C는 skip함
````

## **<font color=darkorange>!</font> 멀티조건 VS 중첩if <font color=darkorange>!</font>**

## **멀티조건문**

````python
age = 25
gen = 'm'
height = 178

if (age < 30 and gen =='m' and height >=150 ) :
    print("군필")
else :
    print("미필")
````

````python
age = 66
gen = 'm'

if (age < 30 and gen == 'm' ) :
    print("군필")
else:
    print("미필")
````

````python
age = 25
gen = 'f'

if (age < 30 and gen == 'm' ) :
    print("군필")
else:
    print("미필")
````

## **중첩if문**
* 조건이 복잡하거나 디테일을 잡고 싶을 때

````python
age = 66
gen = 'm'

if (age < 30) :
    print("나이적합")
    if (gen == 'm') :
        print("군필")
    else :
        print("미필:성별")
else :
    print("미필:나이")
````

````python
age = 25
gen = 'f'

if (age < 30) :
    print("나이적합")
    if (gen == 'm') :
        print("군필")
    else :
        print("미필:성별")
else :
    print("미필:나이")
````
