## **C : 생성**

```python
mylist = pd.Series([1, 3, 5, np.nan, 6, 8])
print(mylist)

s = pd.Series(mylist)
print(s)
```

```python
mylist = [ [1, 3, 5, np.nan, 6, 8] ,
           [1, 3, 5, np.nan, 6, 8] ,
           [1, 3, 5, np.nan, 6, 8] 
         ]

df = pd.DataFrame (mylist)
print(df)
```

```python
mylist = [ [1, 3, 5, np.nan, 6, 8] ,
           [1, 3, 5, np.nan, 6, 8] ,
           [1, 3, 5, np.nan, 6, 8]    # nan = 없는값, 입력/수정 안 되는 진짜 없는 거 (여러 행열 중에 non이 하나라도 있으면 아예 연산 불가)
         ]

df = pd.DataFrame (mylist, index=[10,20,30], columns=['a','b','c','d','e','f'])   # 각 행열에 네이밍
print(df)

df   #print 없이 하면 진짜 표처럼 예쁘게!
```

```python
df.dtypes
```

```python
mydict = { "a":[1,1,1], "b":[3,3,3], "c":[5,5,5] }      #컬럼 네이밍을 abc로 지정해버림 ( i&c 불필요)
df = pd.DataFrame(mydict)
df
```

```python
mylist2 = [ [1, 2, 3],
            [2, 3, 4],
            [3, 4, 5] ]
df = pd.DataFrame(mylist2)
df
```

```python
mydict2 = { "title":"ddd", "a":[2,4,6], "N":["dd", "ss", "ff"] }
df = pd.DataFrame(mydict2)
df
```
