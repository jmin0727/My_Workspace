## **컬렉션**
* <font color=red><b>리스트 : [ ]</b></font>
* <font color=red><b>튜플 : ( ) 생성,조회만 가능(수정 삭제 불가)
* <font color=red><b>딕셔너리 : { } 

```python
mylist = [1, 2, 3, "가", "나"]
print(mylist)
```

```python
mytuple = (1, 2, 3, "가", "나")
print(mytuple)
```

```python
mydict = {"등수":1  ,  "성별":'2'  ,  '등급':'가등급'}
print(mydict)
```

### **C (Create)**

```python
mylist = [1,2,3,"가","나"]
print(mylist)
```

```python
mylist.append("카")
print(mylist)
```

```python
mytuple = (1,2,3,"가","나")
print(mytuple)
```

```python
mydict = {"등수":1  ,  "성별":'2'  ,  '등급':'가등급'}
print(mydict)
```
