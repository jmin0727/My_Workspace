## **[실습] youtube-search-python API 활용**
* pip install youtube-search-python
* https://github.com/alexmercerind/youtube-search-python
* lib proxy 버전 업데이트
     * pip uninstall httpx -y
     * pip install httpx==0.27.2

* 유튜브 API, 제목/url/링크 추출

````python
from youtubesearchpython import VideosSearch
from pprint import pprint
res = VideosSearch('뽀로로', limit = 5)
# pprint(res.result())

# 'title', 'thumbnails'의 'url', 'link' :
for a in res.result()["result"]:            #검색 결과를 하나씩 가져옴 ["검색어"]
    print(a ["title"])                      #영상 제목 출력
    print(a ["thumbnails"] [0] ["url"] )    #썸네일 url 출력   (썸네일의 url, 키값 없이 밸류만)
    print(a ["channel"] ["link"] )          #채널 링크 출력
    print("-"*50)
````

## **[실습] 공공데이터포털**

* https://www.data.go.kr 
* https://dataapi.co.kr/
* pip install requests

> 참고) text = res.content.decode("utf-8-sig")    == 한글인식 코드

````python
import requests   #설치파일, 요청하면 결과를 받겠다는 기능
import json       #python 기본패키지

url = "http://www.gimhae.go.kr/openapi/tour/restaurant.do"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/151.0.0.0 Safari/537.36"
    ),
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Accept-Language": "ko-KR,ko;q=0.9,en-US;q=0.8,en;q=0.7",
    "Referer": "https://www.gimhae.go.kr/",
}


res = requests.get(      #여기가 불러오는거
    url,                 #url, headers 넣으면 결과 받아오겠다
    headers=headers,
    timeout=10           #10초동안 응답이 없으면 그만
)

print(res.status_code)    #HTTP 상태 코드(Status Code) : 처리 결과  (200=성공)

data = res.json()         #.json 형태로 받겠다 (텍스트는 .text())
print(data["results"])    #키값 results만 꺼내겠다
````

* name, category, phone, address 추출

````python
for v in data ["results"] :   #요청한 데이터를 v에 담고
    print(v ["name"], v ["category"], v ["phone"], v ["address"], v ["xposition"], v ["yposition"] )   #각 키값을 차례대로 출력
    print("-"*50)
````
