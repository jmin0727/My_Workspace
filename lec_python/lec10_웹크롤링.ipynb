## **requests 정적 크롤링**

```python
from bs4 import BeautifulSoup
import requests

url = "https://star.ytn.co.kr/news/list.php?mcd=0117&hcd=04"
res = requests.get(url)      #여기가 불러오는거
print(res.status_code)       #200 뜨면 크롤링 가능
#print(res.text)

html_doc = res.text
soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.title)
```

```python
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")    #selelct 다 가져오겠다
for news in news_list : 
    title = news.select_one("div.text_area > div.title > a").text           #1개씩 텍스트로 뽑겠다
    rdate = news.select_one("div.text_area > div.info > div.date").text
    img = news.select_one("div.photo > a > img").get("data-src")            #이미지로 뽑겠다
    href = news.select_one("div.text_area > div.title > a").get("href")      #링크로 뽑겠다
    print(title, rdate, img, href)
```

## **selenium 동적 크롤링**

```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

url = "https://star.ytn.co.kr/news/list.php?mcd=0117&hcd=04"
driver = webdriver.Chrome()               #requests가 아닌 selenium의 webdriver가 크롬브라우저에서 get 해 옴
driver.get(url)

time.sleep(2)  #2초만 기다렸다가 아래 라인들 실행 (게시물 뜨는 시간)

# 3. 더보기 버튼 3번 클릭
for i in range(3):
    btn = driver.find_element(By.CSS_SELECTOR,"a.btn_white_arr_down")    #더보기 버튼의 소스 정보를 찾고(whie~~), 클릭하고, 2초만 기다려라
    btn.click()
    time.sleep(2)
    
# 4. 현재 화면의 HTML 가져오기
html = driver.page_source                    #sekenium을 이용해서 페이지소스를 가져와라 = html이 옴
soup = BeautifulSoup(html, "html.parser")
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")

for news in news_list:
    title = news.select_one("div.text_area > div.title > a").text
    rdate = news.select_one("div.text_area > div.info > div.date").text
    img   = news.select_one("div.photo > a > img").get("data-src")
    href   = news.select_one("div.text_area > div.title > a").get("href")
    print(title, rdate, img, href)

driver.quit()
```

## **[실습] 부천시 맛집 크롤링**

* **정적크롤링**

```python
from bs4 import BeautifulSoup
import requests

url = "https://www.bucheon.go.kr/site/program/board/photoboard/list?boardtypeid=26778&menuid=148006001012001"
res = requests.get(url)      #여기가 불러오는거
print(res.status_code)       #200 뜨면 크롤링 가능
#print(res.text)


html_doc = res.text
soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.title)
```

```python
bc_list = soup.select("#boardForm > div > ul > li > div")   #원본은 #boardForm > div > ul > li:nth-child(1) > div:nth-child(1) li, div의 1번째를 가져오라고 되어있음
for food in bc_list : 
    #title = food.select_one("div > a > img").get("alt")   #a가 두개, alt가 타이틀 속성으로 되어있음
    title = food.select("a")[1].text                       #a가 두개, 1번째 가져와 (여기는 select)
    img = "https://www.bucheon.go.kr" + food.select_one("div > a > img").get("src")  #src 같은 속성 잘 봐야함
    addr = food.select_one("p:nth-child(3)").text
    tel = food.select_one("p:nth-child(4)").text            
    print(title, addr, tel, img)
```

* **동적크롤링**

```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

url = "https://www.bucheon.go.kr/site/program/board/photoboard/list?boardtypeid=26778&menuid=148006001012001"
driver = webdriver.Chrome()
driver.get(url)     #웹페이지 접속
time.sleep(2)


for page in range(1, 4):
    print(f"{page}---------------------------------------------")
    # 2페이지부터 URL에 페이지 번호를 붙여 다음 페이지로 이동
    if page > 1:
        page_url = f"{url}&currentpage={page}"
        driver.get(page_url)
        time.sleep(2)
    html_doc = driver.page_source       # 현재 페이지의 HTML 소스 가져오기

    # requests로 가져온 HTML처럼 BeautifulSoup으로 데이터 추출
    soup = BeautifulSoup(html_doc, "html.parser")
    food_list = soup.select("#boardForm > div > ul > li > div")
    for food in food_list:
        title = food.select_one("div > a > img").get("alt")                                   # 이미지의 alt 속성에서 제목 추출
        #title = food.select("a")[1].text
        img = "https://www.bucheon.go.kr" + food.select_one("div > a > img").get("src")       # 이미지 URL 추출
        addr = food.select_one("p:nth-child(3)").text           # 세 번째 p 태그에서 주소 추출
        tel = food.select_one("p:nth-child(4)").text            # 네 번째 p 태그에서 주소 추출
        print(title, img, addr, tel)
        
    
driver.quit()      #브라우저 종료
```
