#### print 기본값: 줄바꿈 ( print 값or변수, <font color=blue> <u> end="/n" </font> ) 
* <font size=2> 엔드라인 변경 시 아래 참고

````python
for v in range(1, 11) :
    print(v*10)  #엔드라인 기본(줄바꿈)
````

````python
for v in range(1, 11) :
    print(v*10 , end=" ")  #공백(띄어쓰기)
````

````python
for v in range(1, 11) :
    print(v*10 , end="\t")  #탭
````

````python
for v in range(1, 11) :
    print ( v * 10 , end=" " )
    if ( v != 10 ) : 
        print ( "- ", end="" )     # v의 값이 10과 다르다면 하이픈 추가, 줄바꿈 X
    else :
        print( "", end="" )     # v의 값이 10과 다르지 않다면 추가 및 줄바꿈 X
````

### **연습문제**

* range 사용
* 1 ~ 10 짝수면 짝, 홀수면 홀 출력 (숫자도 같이)
* 줄바꿈 X

````python
for number in range(1, 11) :   #1부터 10까지
    if ( number % 2 == 0) :  #숫자를 2로 나눴을 때!! 나머지가 0과 같으면?
        print ( number, "짝수", end=", ")   #짝수, 엔드라인에 쉼표 & 줄바꿈 X
    else :
        print ( number, "홀수", end=", ")   #홀수, 엔드라인에 쉼표 & 줄바꿈 X
````

* 1부터 10까지의 합

````python
total = 0
for number in range(1, 11) :
    total = total + number    #넘버의 합을 토탈(외부)에 대입
print(total)
````

````python
total = 0   #합산할 외부 변수를 만들어줘야 함

for number in range(1, 11) :
    if ( number % 2 == 0) :  #짝수일 때만 
        total = total + number   #값을 토탈에 합하세요 (조건이 위에 있기 때문에 if문 보다 들여써야 함)

print (total) 
````

````python
total = 0

for number in range(1, 11) :
    if ( number % 2 == 1) :  #홀수일 때만 
        total = total + number   #값을 토탈에 합하세요 (조건이 위에 있기 때문에 if문 보다 들여써야 함)

print (total) 
````

````python
total = 0 

for number in range(1, 11) :
    if ( number % 2 != 0) :  #짝수가 아니면! 
        total = total + number   #값을 토탈에 합하세요 (조건이 위에 있기 때문에 if문 보다 들여써야 함)

print (total) 
````

* 리스트에서 20보다 작은 3의 배수를 출력

````python
mylist = [ 13, 21, 12, 14, 30, 18 ]

for a in mylist :
    if ( a < 20  and  a % 3 == 0 ) :
        print ( a )
````

````python
mylist = [ 13, 21, 12, 14, 30, 18 ]

for a in mylist :
    if ( a < 20  and  a % 3 == 0 ) :
        print ( a , end=" ")
````

* 구구단 만들기

````python
for dan in range ( 2, 10 ):
    #print(dan, "단")
    print(f"{dan}단")   #변수와 글자 사이에 공백 없이 출력하고 싶을 떄 
````

````python
for dan in range (2, 10) :
    for gob in range ( 1, 10 ): 
        print(f"{dan*gob}", end="\t")
    print()
````

````python
for dan in range (2, 10) :
    for gob in range ( 1, 10 ): 
        print(f"{dan}*{gob}={dan*gob}", end="\t")
    print()
````

````python
for dan in range (2, 10) :
    print(f"[{dan}단]", end="\t")
    for gob in range ( 1, 10 ): 
        print(f"{dan}*{gob}={dan*gob}", end="\t")
    print()
````

````python
for dan in range (2, 10) :
    print(f"[{dan}단]", end="\t")   #9단까지 루프 끝나면
print()   #프린트

for gob in range ( 1, 10 ):    #1부터 9까지 한 줄이어야 하니까 이거 먼저 하고, 끝나면
    for dan in range (2, 10) :   #2단부터 9단까지
        print(f"{dan}*{gob}={dan*gob}", end="\t")    #9단까지는 탭간격으로, 끝나면
    print()                                          #줄바꿈
````

## **enumerate : 목록의 순서(인덱스)와 값을 함께 꺼내서 반복**

````python
for v in [1,2,3] :
    print(v)
````

````python
for 번째, v in enumerate( [1,2,3] ) :
    print(v, 번째)   #두번째 자리가 인덱스
````
