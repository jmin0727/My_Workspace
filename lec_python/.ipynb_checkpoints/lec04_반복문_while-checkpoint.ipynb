## **while : 조건이 참이면 계속 반복**
* 영원히 반복되기 때문에 <font color=red> break 조건을 잘 걸어야 함 </font>

```python
while 조건문 :
    코드1
    코드2
        break

````python
number = 1
while (True) : 
    print (number)
    number = number + 1

    if (number == 5) :
        break
````

````python
number = 1
while (True) : 
    print (number)
    
    if (number == 5) :
        break
    number = number + 1

#넘버는 무조건 참이라서 무조건 루프됨, 그래서 if문으로 5까지 리미트
````

````python
number = 1
while (number <= 5) : 
    print (number)
    number = number + 1

#넘버를 무조건 참이 아닌 5 이하로 루프, 자동 브레이크됨 (if문으로 리미트 걸지 않아도 됨)
````

````python
number = 2
while (number <= 9) : 
    print (f"{number}단", end="\t")
    number = number + 1
````

````python
number = 2
while (number <= 9) : 
    print (f"{number}단", end="\t")
    number += 1    #number = number + 1  같은 문법
````

### <font color=red> casting : type 변환
* 숫자로 변환하려면 int()

````python
age = input("나이를 입력하세요")
print( int(age)-5)
````

````python
word = input("알파벳을 입력하세요")
if (word in ["Q", "q"]) : 
    print("종료")
else :
    print(f"입력글자 : {word}")
````

````python
while (True) :
    word = input("알파벳을 입력하세요")
    if (word in ["Q", "q"]) : 
        print("종료")
        break      #여기서 꼭 브레이크!
    else :
        print(f"입력글자 : {word}")
````

## **중첩리스트**
* 리스트 안에 리스트 (튜플이나 딕셔너리 다 가능)

#### <font color=blue> **입력**

````python
addr_list = [ ["kim", "010"], ["hong", "000"] ]
print(addr_list)
````

````python
addr_list.append( ["park", "111"] )
addr_list
````

#### <font color=blue> **조회**

````python
len( addr_list )
````

````python
for a in addr_list :
    print(a)             #여기서 a는 리스트 타입
````

````python
for a in addr_list :
    print(a[0], a[1])      #리스트의 0번째, 1번째  이렇게 하면 대괄호 없이 나옴
````

````python
for n, t in addr_list :    #두가지 변수 값 추출하겠다 (n, t)
    print(n, t)            #N번째 말고 그냥 변수로 추출
````

* hong을 찾아라

````python
for n in addr_list :          #n은 addr_list 자체라서 리스트 안에서 첫번째 자리 값을 꺼내야함
    if ( n[0] == "hong" ) :     #리스트의 첫번째가 hong이면!
        print(n[0]) 
````

````python
for n, t in addr_list :        #변수로 구분 후
    if ( n == "hong" ) :        #n이 hong이면!
        print(n) 
````

````python
for n, t in [addr_list[1]] :   #변수를 두개 걸어서 구분해놓고(이름, 번호) 인덱싱으로 1번째 추출
    print(n)
````

#### <font color=blue> **수정**

* hong의 전화번호 수정

````python
for n in addr_list :        #n을 찾고
    if ( n[0] == "hong" ) :   #n의 0번째가 hong이면
        n[1] = 222            #n의 1번째를 111로 바꿔라
````

#### <font color=blue> **삭제**

````python
for n in addr_list :
    if ( n[0] == "hong" ) :
        addr_list.remove(n)
````
