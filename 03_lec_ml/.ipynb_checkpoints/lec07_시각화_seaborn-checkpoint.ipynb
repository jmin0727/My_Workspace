# 기본

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

#인증서 헤제
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

sns.set()    #seaborn 스타일 적용

#--------차트 관련 속성 (한글처리, 그리드)  폰트 설정 가능
plt.rcParams['font.family']= 'Malgun Gothic'

plt.rcParams['axes.unicode_minus'] = False
```

```python
df = sns.load_dataset("penguins")
df
```

```python
sns.pairplot(df, hue="species")
plt.show()
```

# 라인차트 : plt.plot() VS sns.lineplot()

```python
df = sns.load_dataset("fmri")
df.head(3)
```

```python
plt.figure(figsize=(8,2))

sns.lineplot(x="timepoint", y="signal", data=df, hue="region", errorbar=None)
plt.show()
```

# 바차트 : plt.bar() VS sns.barplot()

```python
plt.figure(figsize=(8,2))

sns.barplot(x="timepoint", y="signal", data=df, hue="region", errorbar=None)
plt.show()
```

# 스캐터 : plt.scatter() VS sns.scatterplot

```python
df = sns.load_dataset("diamonds")
df.head(2)
```

```python
plt.figure(figsize=(8,4))

sns.scatterplot(x="carat", y="price", hue="clarity", data=df)
plt.show()
```

# 히스토그램 : plt.hist() VS sns.histplot

```python
plt.figure(figsize=(8,4))

df = sns.load_dataset("diamonds")
sns.histplot(x="price", hue="cut", multiple="stack", data=df)
plt.show()
```

```python
df = sns.load_dataset("penguins")
sns.displot(x="flipper_length_mm", col="species", row="sex", data=df)
plt.show()
```

# 박스플롯 : plt.boxplot() VS sns.boxplot()

```python
df = sns.load_dataset("tips")
df.head()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(x="day", y="total_bill", hue="smoker", data=df)
plt.show()
```

# 바이올린 플롯 : plt.violinplot() VS sns.violinplot()

```python
plt.figure(figsize=(8,4))

sns.violinplot(data=df, x="day", y="total_bill", hue="smoker", split=True, inner="quart")
plt.show()
```

# 히트맵 : sns.heatmap()
* plt.matshow()

```python
df = sns.load_dataset("flights")
df.head(3)
```

```python
dfpv = df.pivot(index="month", columns="year", values="passengers")
dfpv
```

```python
sns.heatmap(dfpv, annot=True, fmt="d")
plt.show()
```
