> plotly : 웹 연동 동적 시각화 <br>
> folium : GIS 지도 시각화 <br>

* <b> 부천역 & 학원 위치 찍기
* location=[37.4848, 126.7831] : 위경도
* tooltip="Click me!" : 풍선도움말
* popup="Timberline Lodge" : 클릭하면 뜨는 오픈창

```python
import folium
from folium import plugins

m = folium.Map(location=(37.48689, 126.78298), zoom_start=18)

folium.Marker(
    location=[37.48689, 126.78298],
    tooltip="Click me!",
    popup="<img width=100 height= 100 src='https://www.wooriac.co.kr/images/common/logo.svg'>우리인재개발원",
    icon=folium.Icon(color="green"),
).add_to(m)


folium.Marker(
    location=[37.484056, 126.782528],
    tooltip="Click me!",
    popup="부천역",
    icon=folium.Icon(color="blue"),
).add_to(m)

folium.CircleMarker([37.484056, 126.782528], radius=100,color='#3186cc', fill_color='#3186cc', popup='부천').add_to(m)

m
```

# **마커클러스터**

```python
import folium
from folium import plugins

N = 100

data = np.array(
[
np.random.uniform(low=36, high=38, size=N), # Random latitudes.
np.random.uniform(low=127, high=128, size=N), # Random longitudes.
] ).T


print(data[:3])
popups = [str(i) for i in range(N)] # Popups texts are simple numbers.

m = folium.Map([37.566345, 126.977893], zoom_start=8)
plugins.MarkerCluster(data, popups=popups).add_to(m)

m
```
