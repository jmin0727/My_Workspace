# **★★★Grouping★★★**

## <font color=blue> **max(), min(), mean(), sum(), count(), std()**

```python
#sample

df77 = df = pd.DataFrame(
    {
        "A": ["one", "one", "two", "three"] * 3,
        "B": ["A", "B", "C"] * 4,
        "C": ["foo", "foo", "foo", "bar", "bar", "bar"] * 2,
        "D": np.random.randn(12),
        "E": np.random.randn(12),
    }
)
df77.head(3)
```

```python
df77.groupby("A")      # 메모리 주소가 찍힘
```

```python
df77.groupby("A")["D"].sum()     #A그룹 중에서 D를 합해라
```

```python
df77.groupby("A")["D"].min()    #최소값
```

```python
df77.groupby("A")["D"].agg(['min', 'max'])    #최소값/최대값을 agg 모으다
```

```python
df77.groupby("A").agg({"D":"min", "E":"max"})    #D는 min, E는 max를 구해라(딕셔너리)
```

##  <font color=blue> **concat & merge**

### **pd.concat(axis=)**
* <font color=red><b> axis=0 : 행 합치기
* <font color=red><b> axis=1 : 열 합치기

```python
dfa = pd.DataFrame ({"a" : [1,2,3], "b" : [10,20,30]})
dfb = pd.DataFrame ({"a" : [4,5], "b" : [40,50]})
```

```python
dfa
```

```python
dfb
```

```python
dfab = pd.concat ( [dfa, dfb],  axis=0 )    #행에 추가, 인덱스 이름 주의
dfab
```

```python
dfab = pd.concat ( [dfa, dfb],  axis=0 ) 
dfab = dfab.reset_index()      #원래 인덱스를 밖으로 꺼내고 인덱스 재정리
dfab
```

```python
dfab = pd.concat ( [dfa, dfb],  axis=0 ) 
dfab = dfab.reset_index(drop=True)      #밖으로 꺼낸 원래 인덱스 지우고 인덱스 재정리
dfab
```

```python
dfab = pd.concat ( [dfa, dfb],  axis=1 )    #열에 추가, 컬럼 이름 주의
dfab
```

```python
dfab = pd.concat ( [dfa, dfb],  axis=1 )
dfab.columns = ['a', 'b', 'c', 'd']           #컬럼 이름 변경
dfab
```

### **pd.merge(dfa, dfb, on="컬럼", how="inner/outer")**
* <font color=red><b> inner : 서로 일치하는 게 있으면 합쳐
* <font color=red><b> outer : 일치하는 게 없으면 비워놔
* concat과의 차이 : 인덱스/컬럼 네이밍 자동

* <font color=blue><b> 컬럼이 아닌 인덱스로 merge

```python
dfa = pd.merge(dfa, dfb, left_index=True, right_index=True, how="inner")      #왼쪽(dfa)인덱스 사용? 오른쪽(dfb)인덱스 사용? / 일치하는게 있으면 합쳐라
dfab
```

```python
dfa = pd.merge(dfa, dfb, left_index=True, right_index=True, how="outer") 
dfab
```

* <font color=blue><b> 컬럼으로 merge

```python
left = pd.DataFrame({"key": ["foo", "foo"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo"], "rval": [4, 5]})
```

```python
left
```

```python
right
```

```python
dfab = pd.merge(left, right, on="key", how="inner")     #foo에 해당하는 게 두개기 때문에 총 네개가 나오는 거
dfab
```

```python
left = pd.DataFrame({"key": ["foo", "foo2"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo2"], "rval": [4, 5]})
```

```python
dfab = pd.merge(left, right, on="key", how="inner")              #foo와 foo2로 구분했기 때문에 총 두개만 나옴
dfab
```
