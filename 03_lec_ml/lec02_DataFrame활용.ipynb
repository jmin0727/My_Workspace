## **필터링**
* emp [조건문] [컬럼]
* emp [컬럼] [조건문]
* 컬럼/조건 둘 중 하나만 있어도 됨
* <font color=red> <b> 조건이 여러개일 때  :  [ (조건절) & | (조건절) ]   괄호 필수
  * <font color=red> <b>& : and   /   | : or   /   ~ : not

> DEPTNO 가 30인 것만 추출

```python
# 조건 : emp["DEPTNO"] == 30
emp [ emp["DEPTNO"] == 30 ]
```

> DEPTNO 가 30인 것 중에서 컬럼 세개만 추출

```python
emp [ emp["DEPTNO"] == 30 ] [[ "EMPNO", "ENAME", "DEPTNO" ]]
```

> DEPTNO가 30이거나 20인 것만 추출

```python
emp [ (emp["DEPTNO"] == 30) | (emp["DEPTNO"] == 20) ]
```

> DEPTNO가 30이거나 20인 것만 추출  (isin 문법)

```python
emp [ emp ["DEPTNO"].isin( [30,20] ) ]
```

> 30이거나 20이 "아닐 때" <br>
> **~ (조건절) : 부정**

```python
emp [ ~ emp ["DEPTNO"].isin( [30,20] ) ]
```

### **실습**

* DEPTNO = 10인 사원 출력

```python
emp [ (emp["DEPTNO"] == 10) ]
```

* DEPTNO = 10, SAL >= 2000 조건의 사원 출력

```python
emp [ (emp["DEPTNO"] == 10) & (emp["SAL"] >= 2000) ]
```

* DEPTNO = 10, SAL >= 2000 조건의 EMPNO, ENAME, SAL, DEPTNO 출력

```python
emp [["EMPNO", "ENAME", "SAL", "DEPTNO"]] [ (emp["DEPTNO"] == 10) & (emp["SAL"] >= 2000) ]
```
