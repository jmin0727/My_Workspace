# 기본

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set()    #seaborn 스타일 적용

#차트 관련 속성 (한글처리, 그리드)  폰트 설정 가능
plt.rcParams['font.family']= 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(10,2))

plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'b--', label="블루")
plt.plot([1, 2, 3, 4], [2, 5, 10, 14], 'r--', label="레드")

plt.fill_between([2,3], [4,9], alpha=0.5)

plt.vlines(2, 5, 15, color='darkgray', linestyle=':', linewidth=2)
plt.hlines(1.8, 3, 4, color='red', linestyle='solid', linewidth=3)

plt.title("차트제목")
plt.xlabel('X-한글')
plt.ylabel('Y-Label')

plt.xlim(0, 5)
plt.ylim(0, 20)

plt.grid(True)

plt.legend(loc="upper right")

plt.show()
```

* **plt.plot( [x], [y] )**
* figure(figsize=(10,2)) : 사이즈
* 포맷 문자열 ‘ro’는 빨간색 (‘red’)의 원형 (‘o’) 마커
* 'b-'는 파란색 (‘blue’)의 실선 (‘-‘)
* Label : 축 이름
* xlim/ylim : 범위
* legend(loc="upper right") : 범례(위치), 범례 이름은 맨위 label="name" (단, 각각 한줄씩 해야함)
* fill_between : 영역채우기 (x, y)
* alpha : 투명도
* vlines(2, 5, 15) : 수직선 (x, y_min, y_max)
* hlines(1.8, 3, 4) : 수평선 (y, x_min, x_max)

# 이중축

```python
fig, ax1 = plt.subplots(figsize=(8,2))
 
ax1.plot([1, 2, 3, 4], [1, 4, 9, 16], color='green')

ax2 = ax1.twinx()
ax2.plot([1, 2, 3, 4], [200, 210, 240, 250], color='deeppink')

plt.show()
```

# 이중차트

```python
fig, ax1 = plt.subplots(figsize=(8,2))
 
ax1.bar([1, 2, 3, 4], [2, 5, 10, 14], color='lightpink')

ax2 = ax1.twinx()
ax2.plot([1, 2, 3, 4], [1, 4, 9, 16], color='skyblue')

plt.show()
```

# plt.subplot<font color=red>s</font>() VS plt.subplot()

* **subplots**

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 3))        #ax1, ax2 = 차트 2개   /   sunplots(1, 2) = 1행 2열  (2,2) = 2행 2열

ax1.bar ([1, 2, 3, 4], [2, 5, 10, 14])
ax2.plot([1, 2, 3, 4], [1, 4, 9, 16]  , color='red')
ax3.bar ([1, 2, 3, 4], [2, 5, 10, 14])

### ax 개별 라벨 
ax1.set_xlabel('X1-한글')
ax2.set_xlabel('X2-한글')
ax3.set_xlabel('X3-한글')

### 전체 타이틀
fig.suptitle('전체 차트 제목')

### 차트 간 여백
plt.tight_layout()

plt.show()
```

* **subplot : ax 없이 여러개 차트**

```python
plt.figure(figsize=(12,4))

plt.title("이중차트")


plt.subplot(1, 3, 1)        # 1행 3열, 그 중 첫번째
plt.bar ([1, 2, 3, 4], [2, 5, 10, 14])
plt.xlabel("111")
plt.ylabel("1-1-1")

plt.subplot(1, 3, 2)        # 1행 3열, 그 중 두번째
plt.bar ([1, 2, 3, 4], [1, 4, 9, 16])
plt.xlabel("222")
plt.ylabel("2-2-2")


plt.subplot(1, 3, 3)        # 1행 3열, 그 중 세번째
df = sns.load_dataset("fmri")    #seaborn 데이터 가져오기
sns.lineplot(x="timepoint", y="signal", data=df, hue="region", errorbar=None)  #seaborn 으로

plt.tight_layout()

plt.show()
```

# **그래프 엔진 변환 예 : DataFrame --> plotly**

```python
import pandas as pd
pd.options.plotting.backend = "plotly"

df = pd.DataFrame(dict(a=[1,3,2], b=[3,2,1]))
fig = df.plot()
fig.update_layout(width=900, height=300, title="차트제목")
```

# 박스플롯 : plt.boxplot()

```python
plt.figure(figsize=(4,3))

np.random.seed(0)
data_a = np.random.normal(0, 2.0, 25)

data_a = np.append(data_a, 12)     #12가 이상치(outlier)
#print(data_a)

plt.boxplot(data_a)
plt.show()

# ------- 사분위값 계산 --------
min_value = np.min(data_a)
q1 = np.percentile(data_a, 25)
median = np.percentile(data_a, 50)
q3 = np.percentile(data_a, 75)
max_value = np.max(data_a)

print("최솟값 :", min_value)
print("25% (Q1) :", q1)
print("50% (중앙값) :", median)
print("75% (Q3) :", q3)
print("최댓값 :", max_value)
```
